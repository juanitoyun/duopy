# 🔁 Recursión
Caso base + progreso + llamada recursiva.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

```python
factorial(5)
```

```python
assert factorial(0) == 1
assert factorial(5) == 120
print('OK')
```
