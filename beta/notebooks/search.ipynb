# 🔎 Búsqueda binaria
Precondición: lista ordenada.

```python
def buscar(a, x):
    i, j = 0, len(a)-1
    while i <= j:
        m = (i+j)//2
        if a[m] == x:
            return m
        if a[m] < x:
            i = m+1
        else:
            j = m-1
    return -1
```

```python
a = [1, 3, 8, 12, 19]
buscar(a, 12)
```

```python
assert buscar(a, 7) == -1
assert buscar(a, 1) == 0
print('OK')
```
